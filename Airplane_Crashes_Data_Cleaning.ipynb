```python
# Question 4
# Missing values in the dataset.
df.isnull().sum() # This will show the number of missing values in each column of the dataset.

df.info() # This will show the data types of each column and the number of non-null values in each column.
```

# Justifications
For the Time column, missing values were set as NaN since time is not used in any of the necessary analyses (Q5-Q10), and therefore imputing a particular value here would create fake information. There were only 20 missing values in the Location column, so I deleted those rows because the location is important for the required region/country split and mapping of fatalities in questions Q8-Q10. There were 18 missing values in the Operator column, which was then imputed as “Unknown” since operator is not very important for the analysis of fatalities, and deleting the rows would reduce the sample unnecessarily. For the Flight # column, all missing values (which were more than 80%, i.e., 4,199 of 5,268) were removed from the dataset because this feature is too unreliable and does not have any analytical use. The Route was left as NaN since it is a descriptive variable that has no role in the calculations. Type had only 27 missing values, which were imputed as the mode (the most common type of aircraft) because the feature is quite complete, and the aircraft type could help in understanding the information better.

```python
# Question 5
fatality_locations = df[['Date', 'Location', 'Aboard', 'Fatalities']]
fatality_locations.head()
```

```python
# Question 6
fatality_locations.loc[fatality_locations['Fatalities'].idxmax(), 'Date']
```

```python
# Question 7
fatality = fatality_locations.copy()

fatality[['Aboard', 'Fatalities']]
no_fatalities = fatality[fatality['Fatalities'] == 0]
print("Number of crashes with no fatalities:", len(no_fatalities))
no_fatalities

print("Mean passengers aboard:", fatality_locations['Aboard'].mean())
print("Mean fatalities:", fatality_locations['Fatalities'].mean())
```

```python
# Question 8
fatality[['Region', 'State/Country']] = fatality['Location'].str.split(',', n=1, expand=True)
fatality.head()
```

```python
# Question 9
top_100 = fatality.sort_values(
    by='Fatalities',
    ascending=False
).head(100)
top_100
```

```python
# Question 10
fbc = (
    fatality.groupby('State/Country')['Fatalities']
    .sum()
    .dropna()
    .sort_values(ascending=False)
    .head(25)
)

fbc

import matplotlib.pyplot as plt

fbc.plot.pie(
    autopct='%1.1f%%',
    figsize=(10, 10)
)

plt.ylabel('')
plt.title('Fatalities by Country and or State - Top 25')
plt.show()
```
