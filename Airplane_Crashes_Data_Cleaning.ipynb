```python

# Determine the number of rows and columns in the dataset.
import pandas
df = pandas.read_csv("Airplane_Crashes_and_Fatalities_Since_1908.csv")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Display the last 75 rows of the dataset.
df.tail(75)
```

```python
# Missing values in the dataset.
df.isnull().sum() # This will show the number of missing values in each column of the dataset.

df.info() # This will show the data types of each column and the number of non-null values in each column.
```

```python
# Question 5
fatality_locations = df[['Date', 'Location', 'Aboard', 'Fatalities']]
fatality_locations.head()
```

```python
# Question 6
fatality_locations.loc[fatality_locations['Fatalities'].idxmax(), 'Date']
```

```python
# Question 7
fatality = fatality_locations.copy()

fatality[['Aboard', 'Fatalities']]
no_fatalities = fatality[fatality['Fatalities'] == 0]
print("Number of crashes with no fatalities:", len(no_fatalities))
no_fatalities

print("Mean passengers aboard:", fatality_locations['Aboard'].mean())
print("Mean fatalities:", fatality_locations['Fatalities'].mean())
```

```python
# Question 8
fatality[['Region', 'State/Country']] = fatality['Location'].str.split(',', n=1, expand=True)
fatality.head()
```

```python
top_100 = fatality.sort_values(
    by='Fatalities',
    ascending=False
).head(100)
top_100
```

```python
fbc = (
    fatality.groupby('State/Country')['Fatalities']
    .sum()
    .dropna()
    .sort_values(ascending=False)
    .head(25)
)

fbc

import matplotlib.pyplot as plt

fbc.plot.pie(
    autopct='%1.1f%%',
    figsize=(10, 10)
)

plt.ylabel('')
plt.title('Fatalities by Country and or State - Top 25')
plt.show()
```
